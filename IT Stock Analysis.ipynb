```python
import yfinance as yf

ticker = yf.Ticker("NVDA") 

# Use .get() to safely retrieve the price. If it's missing, current_price will be None.
current_price = ticker.info.get('currentPrice')

if current_price is not None:
    print(f"NVIDIA Current Price: ${current_price}")
else:
    print("Could not retrieve current price for NVDA.")

historical_data = ticker.history(period="1mo")
```

```python
import matplotlib.pyplot as plt

# The 'historical_data' variable already contains the last month of data
# We'll plot the 'Close' column.
historical_data['Close'].plot(figsize=(10, 6), title='NVDA Closing Price (Last Month)')

# Add labels for better clarity
plt.xlabel("Date")
plt.ylabel("Closing Price (USD)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
import yfinance as yf

tickers = ["AAPL", "IBM", "NVDA"]
price_results = {}

for ticker_symbol in tickers:
    ticker = yf.Ticker(ticker_symbol)
    
    # Safely retrieve the current price
    current_price = ticker.info.get('currentPrice')
    
    # Store the result in a dictionary
    price_results[ticker_symbol] = current_price

print(price_results)
```

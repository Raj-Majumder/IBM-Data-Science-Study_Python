```python
class Coordinate(object):
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def display(self):
        print(f"Position: ({self.x}, {self.y})")
cursor = Coordinate(15, 30)
cursor.display()
```

```python
class Points(object):
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def print_point(self):
        print('x=', self.x, ' y=', self.y)
p2 = Points('Boston', 'Chicago')
p2.y = 'Denver'
p2.print_point()
```

```python
from datetime import datetime, timedelta

now = datetime.now()
future_time = now + timedelta(days=7, hours=3)
print(f"Current time: {now}")
print(f"Time in 7 days and 3 hours: {future_time}")

time_difference = future_time - now
print(f"Time difference: {time_difference}")
```

```python
  from scipy import signal
  import numpy as np

  # A unit impulse at the 0th element in an array of 8 elements
  impulse = signal.unit_impulse(8)
  print(impulse)
  # Output: [1. 0. 0. 0. 0. 0. 0. 0.]

  # An impulse at index 2 in an array of 7 elements
  impulse_shifted = signal.unit_impulse(7, 2)
  print(impulse_shifted)
  # Output: [0. 0. 1. 0. 0. 0. 0.]
```

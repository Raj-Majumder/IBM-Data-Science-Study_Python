```python
import pandas as pd # Used here for creating a clear table at the end

# 1. Define the Class (The Blueprint)
# Class Name: TourismReport
class TourismReport:
    # Class Attribute (Shared by all instances)
    REGION = "Southeast Asia"

    # Constructor Method: __init__ (Sets the initial state/data for each object)
    def __init__(self, country, arrivals_2023_M, arrivals_2024_M):
        # Instance Attributes (Unique to each country object)
        self.country = country
        self.arrivals_2023_M = arrivals_2023_M
        self.arrivals_2024_M = arrivals_2024_M
        self.growth_rate = self.calculate_yoy_growth() # Call a method inside the constructor!

    # Instance Method: calculate_yoy_growth (Behavior)
    def calculate_yoy_growth(self):
        """Calculates the Year-over-Year (YoY) growth rate."""
        # Formula: (New - Old) / Old * 100
        if self.arrivals_2023_M > 0:
            rate = ((self.arrivals_2024_M - self.arrivals_2023_M) / self.arrivals_2023_M) * 100
            # Return the rate rounded to two decimal places
            return round(rate, 2)
        return 0.0

    # Instance Method: get_report (Behavior)
    def get_report(self):
        """Returns a formatted dictionary of the object's data."""
        return {
            'Country': self.country,
            'Region': self.REGION,
            '2024 Arrivals (M)': self.arrivals_2024_M,
            'YoY Growth (%)': self.growth_rate
        }

# --- END OF CLASS DEFINITION ---

# 2. Instantiating Objects (Creating the 'Cookies' from the 'Cookie Cutter')

# Data based on reported 2024 tourism arrivals in Southeast Asia (in Millions)
# Source: [SOUTHEAST ASIA TOURISM PERFORMANCE 2024 RECAP - The Outbox Company](https://the-outbox.com/southeast-asia-tourism-performance-2024-recap/)

thailand = TourismReport("Thailand", 28.2, 35.5)
malaysia = TourismReport("Malaysia", 20.1, 25.0)
vietnam  = TourismReport("Vietnam", 12.6, 17.6)
singapore = TourismReport("Singapore", 13.6, 16.6) # Estimated 2023 for calculation
cambodia = TourismReport("Cambodia", 5.4, 6.7) # Estimated 2023 for calculation

# 3. Interacting with Objects (Calling their Methods)

print(f"--- Analysis for {thailand.country} ---")
print(f"2024 Arrivals: {thailand.arrivals_2024_M} Million")
# Accessing an attribute that was set by a method in the constructor
print(f"YoY Growth Rate: {thailand.growth_rate}%")
print("-" * 30)

print(f"--- Analysis for {vietnam.country} ---")
# Calling the drive() method from the original reading's car example
vietnam.drive = lambda self: print(f"{self.country} is driving growth!")
vietnam.drive(vietnam)
print(f"YoY Growth Rate: {vietnam.growth_rate}%")
print(f"Did {vietnam.country} surpass pre-pandemic recovery levels? {'Yes' if vietnam.growth_rate > 1.0 else 'No'}") # Simple conditional logic
print("-" * 30)


# 4. Displaying All Objects in a Clear, Tabular Format (Great for Jupyter)
# Collect all reports into a list
all_reports = [
    thailand.get_report(),
    malaysia.get_report(),
    vietnam.get_report(),
    singapore.get_report(),
    cambodia.get_report()
]

# Use the pandas library (common in data science) to create a professional table
df = pd.DataFrame(all_reports)

print("\n\n--- Full Tourism Footfall Report (2024 Projections) ---")
print(df.sort_values(by='YoY Growth (%)', ascending=False))
```

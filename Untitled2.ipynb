```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Data structured from the user's text
data = {
    'Start_Year': [2014, 2011, 2007, 2005, 1999, 1998],
    'End_Year': [2017, 2012, 2011, 2007, 2005, 1999],
    'Company_Role': ['IBM - Global Sales Management & Operations', 'Red Hat - OEM Alliance', 'IBM - Client Rep / Alliances Leader', 'Alcatel Lucent - Account Manager', 'GE Healthcare IT', 'BPL Telecom/HAL'],
    'Achievement': ['Global Sales Management & Operations for IBM’s largest revenue portfolio. Awarded for global programs like Focus Accounts Growth by targeted offerings, Attach offerings campaigns for joint GTM strategy.', 'Signed large Gov deal as OEM Alliance. Commercial business up 150%.', 'Multiple recognition for Revenue growth and new client acquisition. Awarded 100% Club (2009) and Top Gun (2011) for achieving annual quota.', 'Account Manager: Expanded Digital voice & data footprint by replacing legacy telephony systems.', 'Multiple awards for direct and Channel Business set up, new market expansions and complex deals. Grew GE healthcare IT business by 500\%+ in the Eastern region (IN).', 'First 1000 port ISDN telephony deal at HAL. Promoted to Sr Sales Engineer at BPL Telecom.'],
    'Award_Flag': [1, 0, 1, 0, 1, 0]
}

df = pd.DataFrame(data)

# Create a combined text for the timeline labels
df['Display_Text'] = df['Company_Role'] + ': ' + df['Achievement']

# Order the DataFrame by Start_Year descending (most recent on the left)
df = df.sort_values(by='Start_Year', ascending=False).reset_index(drop=True)

# --- Visualization for Horizontal Timeline Infographic ---
color_main = '#3E6680'  # Dark blue-gray for the main points
color_highlight = '#80A8C0'  # Lighter blue-gray for awards
color_line = '#D3D3D3'  # Light gray for the connecting line

# *** CHANGE 1: Increased figsize for more horizontal space ***
fig, ax = plt.subplots(figsize=(16, 6))

# Define the coordinates for the timeline
y_coords = [0.1 if i % 2 == 0 else 0.9 for i in range(len(df))]
x_coords = df['Start_Year']

# 1. Draw the horizontal connecting line (the "spine" of the timeline)
ax.axhline(y=0.5, color=color_line, linewidth=1, linestyle='-', zorder=0)

# 2. Draw the markers for each event
for i, row in df.iterrows():
    # Choose color based on the Award_Flag
    marker_color = color_highlight if row['Award_Flag'] == 1 else color_main

    # Draw the main marker (x=year, y=0.5)
    ax.plot(row['Start_Year'], 0.5, marker='o', markersize=10, color=marker_color, zorder=5)

    # Draw the vertical line connecting the spine to the text
    ax.plot([row['Start_Year'], row['Start_Year']], [0.5, y_coords[i]], color=color_line, linestyle='-', linewidth=0.8, zorder=1)

    # Add the text label
    va = 'top' if y_coords[i] < 0.5 else 'bottom'
    # *** CHANGE 2: Increased vertical separation from the line (0.03 to 0.05) ***
    y_pos = y_coords[i] - 0.05 if y_coords[i] < 0.5 else y_coords[i] + 0.05 

    ax.text(
        row['Start_Year'],
        y_pos,
        f"{row['Start_Year']} - {row['End_Year']}\n"
        f"**{row['Company_Role']}**\n"
        f"{row['Achievement']}",
        ha='center',  # Center alignment over the year
        va=va,
        fontsize=8,
        wrap=True,
        color='black'
    )

# 3. Final plot styling
ax.set_title('Career Timeline: Key Achievements & Milestones', fontsize=16, fontweight='bold', pad=20)

# *** CHANGE 3: Increased x-limit padding to widen the plot ***
ax.set_xlim(df['Start_Year'].min() - 2, df['Start_Year'].max() + 2)

# Set y-limits from 0 to 1 to accommodate the alternating labels
ax.set_ylim(0, 1)

ax.set_yticks([])  # Remove y-axis ticks and labels
ax.tick_params(axis='x', top=False, labelsize=10) # Clean up x-axis ticks

# Remove the axes box
ax.spines['top'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['right'].set_visible(False)

# Add a subtle legend for the award highlight
award_patch = plt.scatter([], [], marker='o', s=100, color=color_highlight, label='Award/Recognition Highlight')
main_patch = plt.scatter([], [], marker='o', s=100, color=color_main, label='Key Milestone')

# Adjust legend placement for horizontal layout
ax.legend(
    handles=[award_patch, main_patch],
    loc='upper center',
    ncol=2,
    frameon=False,
    fontsize=9,
    bbox_to_anchor=(0.5, -0.05)
)

plt.tight_layout(rect=[0, 0.1, 1, 1])
plt.show()
```
